```python
from PIL import Image
import os

def check_images(folder_path, files):
    corrupted = []

    for file in files:
        file_path = os.path.join(folder_path, file)

        try:
            with Image.open(file_path) as img:
                img.verify()
        except Exception:
            corrupted.append(file)

    return corrupted
ai_corrupted = check_images(ai_images_path, ai_files)
real_corrupted = check_images(real_images_path, real_files)

print("Corrupted AI images:", len(ai_corrupted))
print("Corrupted Real images:", len(real_corrupted))
```

```python
from PIL import Image
import os
def get_dimensions(folder_path, files):
    dimensions = []
    for file in files:
        file_path = os.path.join(folder_path, file)
        try:
            with Image.open(file_path) as img:
                dimensions.append(img.size)
        except:
            pass
    return dimensions
ai_dimensions = get_dimensions(ai_images_path, ai_files)
real_dimensions = get_dimensions(real_images_path, real_files)
print("AI images - unique dimensions:", len(set(ai_dimensions)))
print("Real images - unique dimensions:", len(set(real_dimensions)))
```

```python
import hashlib
import os

def get_file_hash(file_path):
    with open(file_path, "rb") as f:
        return hashlib.md5(f.read()).hexdigest()

ai_hashes = {}
real_hashes = {}

for file in ai_files:
    path = os.path.join(ai_images_path, file)
    ai_hashes[file] = get_file_hash(path)

for file in real_files:
    path = os.path.join(real_images_path, file)
    real_hashes[file] = get_file_hash(path)

ai_hash_values = set(ai_hashes.values())
real_hash_values = set(real_hashes.values())

duplicates_between_classes = ai_hash_values.intersection(real_hash_values)

print("Duplicate images between AI and Real:", 
      len(duplicates_between_classes))
```

```python
import tensorflow as tf
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from PIL import Image
from sklearn.model_selection import train_test_split

print("TensorFlow:", tf.__version__)
print("NumPy:", np.__version__)
print("Pandas:", pd.__version__)
```

```python
from tensorflow.keras.preprocessing.image import ImageDataGenerator

# 1. Convert numeric labels to string for Keras flow_from_dataframe
train_df['label_str'] = train_df['label'].astype(str)
val_df['label_str'] = val_df['label'].astype(str)
test_df['label_str'] = test_df['label'].astype(str)

# 2. Set image dimensions and batch size
IMG_SIZE = (224, 224)
BATCH_SIZE = 32

# 3. Rescale pixel values from [0, 255] to [0, 1] + add simple data augmentation for training
train_datagen = ImageDataGenerator(
    rescale=1.0/255.0,
    rotation_range=20,
    horizontal_flip=True
)

val_test_datagen = ImageDataGenerator(rescale=1.0/255.0)

# 4. Create generators from DataFrames
train_generator = train_datagen.flow_from_dataframe(
    dataframe=train_df,
    x_col='filepath',
    y_col='label_str',
    target_size=IMG_SIZE,
    class_mode='binary',
    batch_size=BATCH_SIZE,
    shuffle=True
)

val_generator = val_test_datagen.flow_from_dataframe(
    dataframe=val_df,
    x_col='filepath',
    y_col='label_str',
    target_size=IMG_SIZE,
    class_mode='binary',
    batch_size=BATCH_SIZE,
    shuffle=False
)

test_generator = val_test_datagen.flow_from_dataframe(
    dataframe=test_df,
    x_col='filepath',
    y_col='label_str',
    target_size=IMG_SIZE,
    class_mode='binary',
    batch_size=BATCH_SIZE,
    shuffle=False
)
```

```python
import tensorflow as tf
from tensorflow.keras import layers, models

# 1. Define the CNN Architecture
model = models.Sequential([
    # First Convolutional Block
    layers.Conv2D(32, (3, 3), activation='relu', input_shape=(224, 224, 3)),
    layers.MaxPooling2D((2, 2)),
    
    # Second Convolutional Block
    layers.Conv2D(64, (3, 3), activation='relu'),
    layers.MaxPooling2D((2, 2)),
    
    # Third Convolutional Block
    layers.Conv2D(128, (3, 3), activation='relu'),
    layers.MaxPooling2D((2, 2)),
    
    # Classification Head
    layers.Flatten(),
    layers.Dense(128, activation='relu'),
    layers.Dropout(0.5), # Prevents overfitting
    layers.Dense(1, activation='sigmoid') # Binary output: AI (1) or Real (0)
])

# 2. Compile the Model
model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)

# 3. Model Summary
model.summary()
```

```python
# Train for 10 epochs
history = model.fit(
    train_generator,
    epochs=10,
    validation_data=val_generator
)
```

```python
test_loss, test_acc = model.evaluate(test_generator)
print(f"\nTest Accuracy: {test_acc:.4f}")
print(f"Test Loss: {test_loss:.4f}")
```

```python
import matplotlib.pyplot as plt

# Plot Training & Validation Accuracy
plt.figure(figsize=(12, 4))

plt.subplot(1, 2, 1)
plt.plot(history.history['accuracy'], label='Train Accuracy')
plt.plot(history.history['val_accuracy'], label='Val Accuracy')
plt.title('Model Accuracy')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.legend()

# Plot Training & Validation Loss
plt.subplot(1, 2, 2)
plt.plot(history.history['loss'], label='Train Loss')
plt.plot(history.history['val_loss'], label='Val Loss')
plt.title('Model Loss')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend()

plt.show()
```

```python
import tensorflow as tf
from tensorflow.keras.applications import MobileNetV2
from tensorflow.keras import layers, models

# 1. Load pre-trained MobileNetV2 (excluding its top classification layer)
base_model = MobileNetV2(
    input_shape=(224, 224, 3),
    include_top=False,
    weights='imagenet'
)

# 2. Freeze the pre-trained feature extractor
base_model.trainable = False

# 3. Add custom classification head on top
model_tl = models.Sequential([
    base_model,
    layers.GlobalAveragePooling2D(),
    layers.Dropout(0.3),
    layers.Dense(1, activation='sigmoid') # AI (1) vs Real (0)
])

# 4. Compile with a standard Adam optimizer
model_tl.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
    loss='binary_crossentropy',
    metrics=['accuracy']
)

model_tl.summary()
```

```python
# Train Transfer Learning model
history_tl = model_tl.fit(
    train_generator,
    epochs=10,
    validation_data=val_generator
)
```

```python
test_loss_tl, test_acc_tl = model_tl.evaluate(test_generator)
print(f"\nMobileNetV2 Test Accuracy: {test_acc_tl:.4f}")
print(f"MobileNetV2 Test Loss: {test_loss_tl:.4f}")
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 4))

# Accuracy Plot
plt.subplot(1, 2, 1)
plt.plot(history_tl.history['accuracy'], label='Train Accuracy')
plt.plot(history_tl.history['val_accuracy'], label='Val Accuracy')
plt.title('MobileNetV2 - Model Accuracy')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.legend()

# Loss Plot
plt.subplot(1, 2, 2)
plt.plot(history_tl.history['loss'], label='Train Loss')
plt.plot(history_tl.history['val_loss'], label='Val Loss')
plt.title('MobileNetV2 - Model Loss')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend()

plt.show()
```

```python
# Evaluate MobileNetV2 on the Test Set
test_loss_tl, test_acc_tl = model_tl.evaluate(test_generator)

print("\n" + "="*30)
print(f"MobileNetV2 Test Accuracy: {test_acc_tl * 100:.2f}%")
print(f"MobileNetV2 Test Loss:     {test_loss_tl:.4f}")
print("="*30)
```

```python
import tensorflow as tf

# 1. Unfreeze the base model
base_model.trainable = True

# 2. Print total layers in base model
print(f"Total layers in base model: {len(base_model.layers)}")

# 3. Fine-tune from this layer onwards (keep earlier low-level feature layers frozen)
fine_tune_at = 100

# Freeze all layers before fine_tune_at
for layer in base_model.layers[:fine_tune_at]:
    layer.trainable = False

# 4. Recompile with a very small learning rate to avoid destroying pre-trained weights
model_tl.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=1e-5),
    loss='binary_crossentropy',
    metrics=['accuracy']
)

model_tl.summary()
```

```python
# Fine-tune the model
fine_tune_epochs = 10
total_epochs = 10 + fine_tune_epochs

history_fine = model_tl.fit(
    train_generator,
    epochs=total_epochs,
    initial_epoch=history_tl.epoch[-1] + 1,
    validation_data=val_generator
)
```

```python
# Evaluate fine-tuned model on test set
test_loss_fine, test_acc_fine = model_tl.evaluate(test_generator)

print("\n" + "="*40)
print(f"Fine-Tuned Test Accuracy: {test_acc_fine * 100:.2f}%")
print(f"Fine-Tuned Test Loss:     {test_loss_fine:.4f}")
print("="*40)
```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import classification_report, confusion_matrix

# 1. Reset generator to ensure correct ordering
test_generator.reset()

# 2. Get true labels and predicted probabilities
y_true = test_generator.classes
predictions = model_tl.predict(test_generator)
y_pred = (predictions > 0.5).astype(int).reshape(-1)

# 3. Print Classification Report
target_names = ['Real Art (0)', 'AI Art (1)']
print("\n--- Classification Report ---")
print(classification_report(y_true, y_pred, target_names=target_names))

# 4. Plot Confusion Matrix
cm = confusion_matrix(y_true, y_pred)
plt.figure(figsize=(6, 5))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', 
            xticklabels=target_names, yticklabels=target_names)
plt.title('Confusion Matrix')
plt.ylabel('Actual Label')
plt.xlabel('Predicted Label')
plt.show()
```
